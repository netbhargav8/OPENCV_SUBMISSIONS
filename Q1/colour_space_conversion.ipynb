#### **IMPORTING PACKAGES**

```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### **READING THE IMAGE AND CONVERTING TO HSV and RGB**

```python
my_img_bgr = cv2.imread("my_image_low.jpg")
my_img_rgb = cv2.cvtColor(my_img_bgr, cv2.COLOR_BGR2RGB)
my_img_hsv = cv2.cvtColor(my_img_bgr, cv2.COLOR_BGR2HSV)
my_img_gray = cv2.cvtColor(my_img_bgr, cv2.COLOR_BGR2GRAY)
plt.figure(figsize=[30,10])
plt.subplot(141);plt.imshow(my_img_rgb);plt.title('RGB');plt.axis('off')
plt.subplot(142);plt.imshow(my_img_bgr);plt.title('BGR');plt.axis('off')
plt.subplot(143);plt.imshow(my_img_hsv);plt.title('HSV');plt.axis('off')
plt.subplot(144);plt.imshow(my_img_gray);plt.title('gray');plt.axis('off')
```

Detail Preservation: Look closely at your output. Typically, the Value (V) channel (which represents brightness/luminance independently of color information) or the Saturation (S) channel preserves structural details best in low light.

Why? Human eyes and sensors struggle to perceive color (Hue) in low light (due to low signal-to-noise ratio in color bins), but intensity changes (Value) still capture physical boundaries and edges.
