
#### **1. Define paths for input video clip and output processed video**


```python
import cv2
input_path = 'my_video.mp4'
output_path = 'output.avi'
```


#### **2. Open the video capture object and extract native video properties**

```python
cap = cv2.VideoCapture(input_path)

# Extract native video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"--- [TASK 9 VIDEO PROCESSING INPUT STATS] ---")
print(f"Resolution: {width}x{height} | FPS: {fps} | Total Frames: {total_frames}")
print("----------------------------------------------")
```


#### **3. Setup VideoWriter (Using 'mp4v' codec container)**

```python
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(output_path, fourcc, fps if fps > 0 else 30, (width, height))
```

#### **4.Frame by Frame Processing Loop**

```python
frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break  # End of video stream

    # --- ALL PROCESSING LOGIC BELONGS OUTSIDE THE 'if not ret' CHECK ---
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    edges = cv2.Canny(blurred, 50, 150)

    # BUG FIX: Convert single-channel binary edge map back to 3-channel BGR
    # to prevent format mismatch errors in cv2.VideoWriter
    processed_frame = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

    # Write the processed frame to output file
    out.write(processed_frame)
    frame_count += 1

# Clean up system resources
cap.release()
out.release()
cv2.destroyAllWindows()
print(f"Successfully processed and saved {frame_count} frames to '{output_path}'")
print("--------------------------")
```
