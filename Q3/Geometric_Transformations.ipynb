#### **Import Libraries**

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.core.pylabtools import figsize
```

#### **Image Ingestion and Dimensions**

```python
img_objs = cv2.imread("pen_image.jpg")
h, w = img_objs.shape[:2]
```

#### **Translational(Spatial Shifting)**

```python
M_translational = np.float32([[1, 0, 50], [0, 1, 30]])
img_translated = cv2.warpAffine(img_objs, M_translational, (w, h))
```

#### **Combined Rotation and Scale**

```python
center = (w//2,h//2)
angle = 138.0
scale = 1.6
M_rotate_scale = cv2.getRotationMatrix2D(center, angle, scale)
img_rotated_scaled = cv2.warpAffine(img_objs, M_rotate_scale, (w, h))
```

#### **General Affine Transformation**

```python
pts_src = np.float32([[50, 50], [200, 50], [50, 200]])
pts_dst = np.float32([[70, 100], [220, 80], [100, 250]])

M_affine = cv2.getAffineTransform(pts_src, pts_dst)
img_affine = cv2.warpAffine(img_objs, M_affine, (w, h))
```

#### **Matplotlib Color ReIndexing and Plotting**

```python
plt.figure(figsize=[20,5])
plt.subplot(141);plt.imshow(cv2.cvtColor(img_objs, cv2.COLOR_BGR2RGB));plt.axis('off');
plt.subplot(142);plt.imshow(cv2.cvtColor(img_translated, cv2.COLOR_BGR2RGB));plt.axis('off');
plt.subplot(143);plt.imshow(cv2.cvtColor(img_rotated_scaled, cv2.COLOR_BGR2RGB));plt.axis('off');
plt.subplot(144);plt.imshow(cv2.cvtColor(img_affine, cv2.COLOR_BGR2RGB));plt.axis('off');
```
